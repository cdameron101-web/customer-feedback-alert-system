```python
feedback_data = [
    {
        "customer": "Jordan",
        "rating": 5,
        "feedback": "Amazing service. Everything worked perfectly!"
    },
    {
        "customer": "Maya",
        "rating": 2,
        "feedback": "My order arrived late and customer service never responded."
    },
    {
        "customer": "Alex",
        "rating": 1,
        "feedback": "I was charged twice! I need a refund immediately."
    },
    {
        "customer": "Taylor",
        "rating": 4,
        "feedback": "Good experience overall, but shipping was a little slow."
    },
    {
        "customer": "Morgan",
        "rating": 1,
        "feedback": "My account was hacked and I see unauthorized charges!"
    },
    {
        "customer": "Sam",
        "rating": 3,
        "feedback": "The product is okay, but the instructions were confusing."
    }
]

print(f"Feedback records loaded: {len(feedback_data)}")
```

```python
critical_keywords = [
    "hacked",
    "unauthorized",
    "charged twice",
    "fraud"
]

warning_keywords = [
    "refund",
    "late",
    "never responded",
    "broken"
]

print("Critical keywords:", critical_keywords)
print("Warning keywords:", warning_keywords)
```

```python
critical_keywords = [
    "hacked",
    "unauthorized",
    "charged twice",
    "fraud"
]

warning_keywords = [
    "refund",
    "late",
    "never responded",
    "broken"
]

print("Critical keywords:", critical_keywords)
print("Warning keywords:", warning_keywords)
```

```python
alerts = []

for record in feedback_data:
    customer = record["customer"]
    rating = record["rating"]
    feedback = record["feedback"]

    # Convert feedback to lowercase for easier matching
    feedback_lower = feedback.lower()

    # Look for critical keywords
    critical_matches = [
        word for word in critical_keywords
        if word in feedback_lower
    ]

    # Look for warning keywords
    warning_matches = [
        word for word in warning_keywords
        if word in feedback_lower
    ]

    # Determine severity
    if critical_matches:
        severity = "CRITICAL"
    elif rating <= 2 or warning_matches:
        severity = "WARNING"
    else:
        severity = "OK"

    alerts.append({
        "customer": customer,
        "rating": rating,
        "severity": severity,
        "feedback": feedback
    })

for alert in alerts:
    print(
        f"{alert['customer']}: "
        f"{alert['severity']} "
        f"(Rating: {alert['rating']})"
    )
```

```python
action_required = [
    alert for alert in alerts
    if alert["severity"] != "OK"
]

print(f"Customers requiring attention: {len(action_required)}\n")

for alert in action_required:
    print(
        f"{alert['severity']}: {alert['customer']} "
        f"(Rating: {alert['rating']})"
    )
    print(f"Feedback: {alert['feedback']}")
    print("-" * 50)
```

```python
report = "CUSTOMER FEEDBACK ALERT REPORT\n"
report += "=" * 40 + "\n\n"
report += f"Customers requiring attention: {len(action_required)}\n\n"

for alert in action_required:
    report += (
        f"{alert['severity']}: {alert['customer']}\n"
        f"Rating: {alert['rating']}/5\n"
        f"Feedback: {alert['feedback']}\n"
        + "-" * 40 + "\n"
    )

with open("customer_feedback_alerts.txt", "w") as file:
    file.write(report)

print(report)
```
